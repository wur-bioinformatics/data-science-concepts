### Plotting
`matplotlib` has several plotting functions. One commonly used function is `plot`, which allows you to plot 2D data. Here is a simple example:

```python
# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 3 * np.pi, 0.1) #here we're using numpy to create the data for the plot on the x axis
y = np.sin(x) #again, numpy is used for the y axis

# Plot the points using matplotlib
plt.plot(x, y)
```

With just a little bit of extra work we can easily plot multiple lines at once, and add a title, legend, and axis labels:

```python
y_sin = np.sin(x)
y_cos = np.cos(x)

# Plot the points using matplotlib
plt.plot(x, y_sin)
plt.plot(x, y_cos)
plt.xlabel('x axis label') 
plt.ylabel('y axis label')
plt.title('Sine and Cosine')
plt.legend(['Sine', 'Cosine'])
```

With subplots, you can plot different things in the same figure using the `subplots` function of Matplotlib. Here is an example:

```python
# Set up a subplot grid that has 1 row and 2 columns,
# and set the figure size.
fig, ax = plt.subplots(1, 2, figsize=(12,5))

# Make A first plot (scatter plot)
ax[1].scatter(df.C, df.D)
ax[1].set_xlabel('Column C')
ax[1].set_ylabel('Column D')
ax[1].set_title('Columns C and D');

# Make A second plot (pie chart)
ax[0].pie(df.A, autopct='%1.2f%%') #'autopct='%1.2f%%' embeds the column A values into the pie slices'
ax[0].set_xlabel('Column A values')
ax[0].set_title('Column A values');
```

# Part 4: Plotnine and the Grammar of Graphics

`Plotnine` is an implementation of a `grammar of graphics` in Python based on `ggplot2`. 

`ggplot2` is a well-known and widely used `R library`. It's involves using a layered approach to create a data plot:

Layer 1 - the data (e.g., a dataframe)
Layer 2 - the axes based on the dimensions (i.e., which columns do you want to plot?)
Layer 3 - the scaling for interpretability (not always needed)
Layer 4 - Geoms (plot types)
Layer 5 - Statsitics (if needed)
Layer 6 - facets for sub-plots, which might be needed for better understanding and insight.

The grammar setup allows you to compose plots by explicitly mapping variables in a dataframe to the visual objects that make up the plot.

Plotting with a grammar of graphics is powerful for communication and allows you to work in a structured way. Custom (and otherwise complex) plots are easy to think about and build incremently, while the simple plots remain simple to create.

```python
import plotnine
from plotnine import ggplot, geom_point, aes, stat_smooth, facet_wrap, geom_tile, geom_text, geom_segment
```

```python
#below we use the df for the dataset, then we are plotting the Year by the Gpa, with the colouring provided by the Year.
#the statisics are lm, which means linear model
#the facet wrap splits the plot into multiple plots by gender.
(ggplot(df, aes("Year", "Gpa", color="factor(Age)")) ##layers 1, 2 and 3
 + geom_point() ## layer 4
 + stat_smooth(method="lm") ## layer 5
 + facet_wrap("~Gender")) ## layer 6
```

However, as a final exercise, let's consider a question we want to investigate. Using the UniComplaints dataset, we would like to determine the Gpa performance by country to understand how students from different background perfrom within the university. 

To do this, we could start by plotting the Gpa along the x-axis and the country on the y-axis, right?

```python
#here we do a segment plot of the Gpa and the nationality
(ggplot()
 # Range strip
 + geom_segment(
     df,
     aes(x='Gpa', xend='Gpa', y='Nationality', yend='Nationality'),
     size=6,
     color='#a7a9ac')

+ geom_point(
     df,
     aes('Gpa', 'Nationality', color='Gpa', fill='Gpa'),
     size=5,
     stroke=0.7)
)
```

The plot is interesting, but not very insightful and we certainly can't use if for data storytelling or capturing anyone's interest. What is missing is the data pre-processing, to make the data more suitable for a plot that we can communicate.

What we need to do is create some new features for the plot- i.e., a subset which contains the max and min value per country. For this we use pandas and numpy.

```python
#we can find out the overall min and max values in Gpa and store it as a variable
res = df['Gpa'].agg(['min', 'max'])
```

```python
res
```

```python
#to store the min max value by nationality, we create a new variable called min_max
#group by nationality, then use the agg() function to calculate the mean, min and max
min_max = df.groupby('Nationality', as_index=False).agg({'Gpa': ['mean', 'min', 'max']})#as index = false creates a new column rather than using the countries and index labels
```

```python
min_max
```

```python
stats = pd.DataFrame().assign(Nationality=min_max['Nationality'], Min=min_max['Gpa']['min'], Max=min_max['Gpa']['max'], Mean=min_max['Gpa']['mean'],)
```

```python
stats
```

```python
# Plot the points using matplotlib
a = stats['Min']#blue line
b = stats['Max']#orange line
c = stats['Mean']#green line
x = stats['Nationality'] #again, numpy is used for the y axis

plt.plot(x, a)
plt.plot(x, b)
plt.plot(x, c)

plt.show()
```

```python
#maybe a further inspection using plotly
px.bar(stats, x= 'Nationality', y="Mean", color="Mean", title="Mean per Nationality")
```

```python
(ggplot()

+ geom_segment(stats, aes(x='Min', xend='Max', y='Nationality', yend='Nationality'), size=6, color='#a7a9ac')

+ geom_point(stats, aes('Mean', 'Nationality', color='Nationality', fill='Nationality'), size=5, stroke=0.7) 
 
+ geom_text(stats, aes(x='Max', y='Nationality', label='Max'), size=9, fontweight='bold', format_string='+{}')
 
+ geom_text(stats, aes(x='Min', y='Nationality', label='Min'), size=9, fontweight='bold', format_string='+{}')
 
)
```
